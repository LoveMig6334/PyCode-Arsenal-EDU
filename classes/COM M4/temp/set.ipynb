### Update


```python
# U : Update
set1.update([11, 10])
print(set1)
set2.update(["ประจวบ", "สุพรรณ"])
print(set2)
# เป็นการเพิ่มสมาชิกเข้าไปในsetโดยตรงที่ได้มากกว่าหนึ่ง ต้องใช้[]ครอบสมาชิกที่ต้องการเพิ่มด้วย
```

```python
# R : Remove
set1.remove(4)
print(set1)
# ใช้ลบสมาชิกในsetที่ต้องการ หากไม่มีจำนวนสมาชิกที่ต้องการก็จะทำให้code error
```

```python
# D: Discard
set1.discard(60)
print(set1)
# ใช้ลบสมาชิกใน setที่ต้องการ หากไม่มีจำนวนสมาชิกที่ต้องการก็จะไม่เกิดอะไรขึ้น
```

```python
# P : Pop
set1.pop()
set1
# ใช้ในการลบสมาชิกตัวเเรกของset หากกดruncodeไปเรื่อยๆ จำนวนสมาชิกก็จะค่อยๆลดลงไปเรื่อยๆจากซ้ายไปขวา
```

```python
# C : Clear
set1.clear()
print(set1)
# ใช้ในการลบสมาชิกทั้งหมดในset
```

### Set Method


```python
# U : Union
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

set3 = set1.union(set2)
set4 = set1 | set2
print(set3)
print(set4)
# คำสั่ง union กับ | มีหน้าที่เดียวกันคือการจับset 2 setมารวมสมาชิกกัน
```

```python
# I : Intersection
set3 = set1.intersection(set2)
set4 = set1 & set2
print(set3)
print(set4)
# คำสั่ง intersection กับ & มีหน้าที่เดียวกันคือการจับset 2 setมารวมสมาชิกที่มีค่าเท่ากันในเเต่ละset
```

```python
# D : Difference
set3 = set1.difference(set2)
set4 = set2 - set1
print(set3)
print(set4)
# คำสั่ง difference กับ - มีหน้าที่เดียวกันคือตั้งsetตัวหนึ่งเป็นหลักเเล้วลบสมาชิกที่เหมือนกันกับอีกsetหนึ่งเเล้วเอาตัวที่เหลือมาเป็นผลลัพธ์
```

```python
# S : Symmetric_difference
set3 = set1.symmetric_difference(set2)
set4 = set1 ^ set2
print(set3)
print(set4)
# คำสั่ง symmetric_difference กับ ^ มีหน้าที่เดียวกันคือการจับ set 2 setมารวมสมาชิกที่มีค่าไม่เท่ากันในเเต่ละset
```

### Subset


```python
# I : Issubset
set1.issubset(set2)
# ใช้ในการตรวจสอบว่าสมาชิกsetหนึ่งอยู่ในสมาชิกของอีกsetหนึ่งมั้ย
```

```python
# I : Issuperset
set1.issuperset(set2)
# ใช้ในการตรวจสอบว่าsetหนึ่งมีสมาชิกทุกตัวของอีกsetหนึ่งอยู่หรือไม่
```

```python
# I : Isdisjoint
set1 = {1, 2, 3, 4}
set2 = {1, 4}
set3 = {1, 5}
set4 = {8, 9}
print(set2.isdisjoint(set1))
print(set3.isdisjoint(set1))
print(set4.isdisjoint(set1))
# ใช้ในการตรวจสอบว่าsetหนึ่งมีไม่มีสมาชิกบางตัวของอีกsetหนึ่งอยู่หรือไม่
```

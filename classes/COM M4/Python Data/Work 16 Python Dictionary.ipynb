```python
# C -> Create Dict

name_dcit = {
    "ID": "11231",
    "Name": "Thatt",
    "SName": "Bunnag",
}
```

```python
name_dcit
```

```python
# R -> Read Dict value

print(name_dcit["ID"])
print(name_dcit["Name"])
print(name_dcit["SName"])
```

```python
# U -> Update Dict

name_dcit["job"] = "student"

name_dcit.update({"sex": "M", "age": 15})
```

```python
name_dcit
```

```python
# R -> Read Dict value

print(name_dcit.get("ID"))
print(name_dcit.get("address", "ไม่พบที่อยู่"))
```

```python
# Deleat Item from dict

name_dcit.pop("ID")
del name_dcit["age"]
```

```python
name_dcit
```

```python
# Addon

print(len(name_dcit))
"name" in name_dcit
```

```python
for i in name_dcit:
    print(i)
```

```python
for u in name_dcit.keys():
    print(u)
```

```python
for v in name_dcit.values():
    print(v)
```

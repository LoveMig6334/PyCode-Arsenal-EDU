```python
# C - Create tuple
my_tuple = (1, 2, 3, 4, 5)
```

```python
# R - Read tuple

print(my_tuple[0])
print(my_tuple[1])
print(my_tuple[2])
```

```python
# U - Update tuple index object

try:
    my_tuple[1] = 69
except Exception as e:
    print(e)

# ไม่สามารถเปลียนแปลง tuple ได้
```

```python
try:
    my_tuple.append(69)
except Exception as e:
    print(e)

# ไม่สามารถเพิ่มค่าลง tuple ได้
```

```python
try:
    my_tuple.pop(2)
except Exception as e:
    print(e)

# ไม่สามารถเพิ่มค่าลง tuple ได้
```

```python
# D - delete tuple

del my_tuple
```

```python
try:
    print(my_tuple)
except:
    print("tuple ถูกลบแล้ว")
```

```python
new_tuple = (1, 2, 3, 4, 5)
```

```python
# Function เสริมใน Tuple

print(len(new_tuple))
print(5 in new_tuple)
```

```python
for member in new_tuple:
    print(member)
```

```python
# การสลับค่าตัวแปรโดยใช้ tuple
# นำ a ไปใส่ใน b และนำ b ไปใส่ใน a

a = 5
b = 10

a, b = b, a

print(f"{a=} {b=}")
```

```python
# Tuple unpack
student_info = ("Mig", 16, 172, 51)
print(student_info)

name, age, hight, waight = student_info

print(f"{name=}")
print(f"{age=}")
print(f"{hight=}")
print(f"{waight=}")
```

```python
# การส่งข้อมูลออกจาก function


def example_fn(r: float) -> tuple:
    pi = 3.14159265
    A = pi * r * r

    return round(A, 2), r


print(example_fn(4))

a, r = example_fn(4)
print(f"{a=} {r=}")
```

```python
list1 = [1, 1, 2, 2, 3, 3, 4, 4]
set1 = {1, 1, 2, 2, 3, 3, 4, 4}
tuple1 = (1, 1, 2, 2, 3, 3, 4, 4)
dict1 = {"a": 1, "b": 2, "c": 3, "d": 4}
print(list1)
print(set1)
print(tuple1)
print(dict1)
```

```python
# list
list_set = list(set1)
list_tuple = list(tuple1)
list_dict_keys = list(dict1.keys())
list_dict_values = list(dict1.values())
list_dict_items = list(dict1.items())
list_dict = list(dict1)
print(list_set)
print(list_tuple)
print(list_dict)
print(list_dict_keys)
print(list_dict_values)
print(list_dict_items)
# เป็นตัวเเปรที่สร้างขึ้นใหม่ สร้างจากตัวฝังซ้ายปลายทาง ตัวฝังขวาต้นทฃาง
# list_dict_itemsจะเป็นlistที่มีtupleภายใน
```

```python
# set
set_list = set(list1)
set_tuple = set(tuple1)
set_dict = set(dict1)
set_dict_values = set(dict1.values())
set_dict_items = set(dict1.items())
print(set_list)
print(set_tuple)
print(set_dict)
print(set_dict_values)
print(set_dict_items)
# set_dictจะไม่ได้เรียงข้อมูล
```

```python
# tuple
tuple_list = tuple(list1)
tuple_set = tuple(set1)
tuple_dict = tuple(dict1)
tuple_dict_values = tuple(dict1.values())
tuple_dict_items = tuple(dict1.items())
print(tuple_list)
print(tuple_set)
print(tuple_dict)
print(tuple_dict_values)
print(tuple_dict_items)
```

```python
# dictionary
# dict_list = dict(list1)
# dict_set = dict(set1)
# dict_tuple = dict(tuple1)
print(dict_list)
print(dict_set)
print(dict_tuple)
# dictไม่สามารถเเปลงจากอะไรได้เลย เพราะdictต้องการค่าสองค่าคือkeyเเละvalue เเต่list set tupleมีค่าข้อมูลเเบบหนึ่งมิติซึ่งdictต้องการสองมิติ
```

```python
list2 = [["a", 1], ["b", 2], ["c", 3]]
set2 = {("a", 1), ("b", 2), ("c", 3)}
tuple2 = (["a", 1], ["b", 2], ["c", 3])

dict_list = dict(list2)
dict_set = dict(set2)
dict_tuple = dict(tuple2)

print(dict_list)
print(dict_set)
print(dict_tuple)
# ถ้าทำlist tuple เป็นสองมิติเเล้วสามารถนำไปใช้กับdictได้ ยกเว้นset
# ถ้าทำsetสองมิติภายในเป็นtupleจะะสามารถใช้กับdictได้
# listสามารถใช้กับtupleได้ tupleสามารถใช้กับlistได้ มีเพียงsetที่ใช้ได้กับเเค่tuple
```

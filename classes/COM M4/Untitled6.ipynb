```python
import random

rand = random.random()
print(rand)
print(int(rand * 20) + 1)  # 0-19
# หากต้องการค่าถึงเท่าไหร่ให้คูณจำนวนนั้นกับ randเเละค่าจะrandomไม่เกินจำนวน
```

```python
odd = 0
event = 0
count = 0
while count <= 1000:
    if random.randint(1, 2) == 1:
        odd += 1  # >>> old = old + 1
    else:
        event += 1
    count += 1
print(odd, event)
```

```python
my_list = [1, 2, 3, 4, 5, "hello", "world"]
my_set = {1, 2, 3, 4, 5, 6}
my_tuple = (10, 20, 30, "python", "copilot")

my_dict = {
    "name": "John",
    "age": 25,
    "city": "New York",
    "skills": ["Python", "JavaScript"],
}

# Print all data
print("List:", my_list)
print("Set:", my_set)
print("Tuple:", my_tuple)
print("Dictionary:", my_dict)
```

# Convert type to Dict


```python
# Create list/tuple of tuples for conversion examples
pairs_list = [("a", 1), ("b", 2), ("c", 3)]
pairs_tuple = (("x", 10), ("y", 20), ("z", 30))

list_to_dict = dict(pairs_list)
print("Converted List of Tuples to Dict:", list_to_dict)

tuple_to_dict = dict(pairs_tuple)
print("Converted Tuple of Tuples to Dict:", tuple_to_dict)


# Converting from zip (creates key-value pairs)
keys = ["name", "age", "city"]
values = ["Alice", 30, "Paris"]
zip_to_dict = dict(zip(keys, values))


print("Converted Zip to Dict:", zip_to_dict)
```

# Convert type to Set


```python
list_to_set = set(my_list)
print("Converted List to Set:", list_to_set)

tuple_to_set = set(my_tuple)
print("Converted Tuple to Set:", tuple_to_set)


dict_to_set = set(my_dict)
dict_key_to_set = set(my_dict.keys())

try:
    dict_value_to_set = set(my_dict.values())
except TypeError:
    dict_value_to_set = "Cannot convert dict values to set due to unhashable types."
# Note: dict.items() returns tuples, which are hashable and can not be converted to set


print("Converted Dict to Set of Keys:", dict_to_set)
print("Converted Dict Keys to Set:", dict_key_to_set)
print("Converted Dict Values to Set:", dict_value_to_set)
```

# Convert type to Tuple


```python
list_to_tuple = tuple(my_list)
print("Converted List to Tuple:", list_to_tuple)

set_to_tuple = tuple(my_set)
print("Converted Set to Tuple:", set_to_tuple)


dict_to_tuple = tuple(my_dict)
dict_key_to_tuple = tuple(my_dict.keys())
dict_value_to_tuple = tuple(my_dict.values())
dict_item_to_tuple = tuple(my_dict.items())


print("Converted Dict to Tuple of Keys:", dict_to_tuple)
print("Converted Dict Keys to Tuple:", dict_key_to_tuple)
print("Converted Dict Values to Tuple:", dict_value_to_tuple)
print("Converted Dict Items to Tuple:", dict_item_to_tuple)
```

# Convert type to List


```python
set_to_list = list(my_set)
print("Converted Set to List:", set_to_list)

tuple_to_list = list(my_tuple)
print("Converted Tuple to List:", tuple_to_list)


dict_to_list = list(my_dict)
dict_key_to_list = list(my_dict.keys())
dict_value_to_list = list(my_dict.values())
dict_item_to_list = list(my_dict.items())


print("Converted Dict to List of Keys:", dict_to_list)
print("Converted Dict Keys to List:", dict_key_to_list)
print("Converted Dict Values to List:", dict_value_to_list)
print("Converted Dict Items to List:", dict_item_to_list)
```
